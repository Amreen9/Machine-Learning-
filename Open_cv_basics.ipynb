```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img=cv2.imread("/content/image.jfif")
```

```python
plt.imshow(img)
```

```python
img2=cv2.imread("/content/image2.jfif")
plt.imshow(img2)
```

```python
height, width, _ = img.shape
img2_resized = cv2.resize(img2, (width, height))
result = img + img2_resized
```

```python
plt.imshow(result)
```

```python
res2=img-img2_resized
plt.imshow(res2)
```

```python
res3=img*img2_resized
plt.imshow(res3)
```

```python
res4=img/img2_resized
plt.imshow(res4)
```

```python
image = cv2.imread("/content/image3.jfif")
image_blur = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
figure_size = 7
results = cv2.blur(image_blur, (figure_size, figure_size))

plt.figure(figsize=(11, 6))
plt.subplot(121)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original")
plt.xticks([]), plt.yticks([])

plt.subplot(122)
plt.imshow(cv2.cvtColor(results, cv2.COLOR_HSV2BGR))
plt.title("Blurred")
plt.xticks([]), plt.yticks([])

plt.show()
```
