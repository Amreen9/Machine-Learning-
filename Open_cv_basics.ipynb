```python
image = cv2.imread("/content/image3.jfif")
image_blur = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
figure_size = 7
results = cv2.blur(image_blur, (figure_size, figure_size))

plt.figure(figsize=(11, 6))
plt.subplot(121)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original")
plt.xticks([]), plt.yticks([])

plt.subplot(122)
plt.imshow(cv2.cvtColor(results, cv2.COLOR_HSV2BGR))
plt.title("Blurred")
plt.xticks([]), plt.yticks([])

plt.show()
```

```python
#zoom
image = cv2.imread("/content/image3.jfif")
(height, width) = image.shape[:2]
zoom=cv2.resize(image,None,fx=2,fy=2,interpolation=cv2.INTER_LINEAR)
plt.imshow(zoom)
```

```python
zoom=cv2.resize(image,(int(width*2),int(height*2)),interpolation=cv2.INTER_CUBIC)
plt.imshow(zoom)
```

```python
#rotate
m=cv2.getRotationMatrix2D((width/2,height/2),45,1)
rotated=cv2.warpAffine(image,m,(width,height))
plt.imshow(rotated)
```

```python
#Translate
m=np.float32([[1,0,100],[0,1,100]])
translated=cv2.warpAffine(image,m,(width,height))
plt.imshow(translated)
```
