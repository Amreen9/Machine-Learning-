Ans: Machine learning is the technique of teaching computers to learn from data so they can make predictions or discover patterns without hard-coded rules

Machine learning is a part of AI, and it helps systems learn from data instead of being programmed with every rule. So, AI is the broader field, while machine learning is one important method used to build AI systems.

**ML and DL**

Deep learning is a subset of machine learning that uses multi-layer neural networks to learn patterns from data.It is especially useful for complex tasks like image recognition, speech processing, and natural language understanding.

***Basics***

- It is inspired by how the human brain processes information.

- It learns by adjusting weights through training, usually with methods like forward propagation and backpropagation.

- It works best with large amounts of data and computing power.

**1. Supervised Learning:**
In supervised learning, the model learns from data that already has the correct answer. It is like learning with a teacher

The two main types of supervised learning are:

*1.Regression*: used to predict a continuous value, like house price or temperature.

*2.Classification*: used to predict a category or class, like spam or not spam, or pass or fail.


***Easy explanation***
- In regression, the answer is a number.
- In classification, the answer is a label or group

**2. Unsupervised Learning:**
In unsupervised learning, the data has no labels or answers. The model tries to find hidden patterns by itself.

The main types of unsupervised learning are:

*1. Clustering:* groups similar data points together, like customer groups

*2. Dimensionality reduction:* reduces the number of features while keeping important information, like PCA.

*3. Association rule learning:* finds items or events that often occur together, like market basket analysis.

*4. Anomaly detection:* finds unusual data points that do not fit the normal pattern scikit-learn.


**3. Reinforcement Learning:**
In reinforcement learning, a system learns by trial and error. It gets rewards for good actions and penalties for bad actions

*1.Self-training:* the model labels some unlabeled data by itself and learns from the confident predictions.

*2.Co-training:* two models help label data for each other geeksforgeeks.

*3.Graph-based methods:* labels spread through data points that are connected or similar.

**4. Semi-supervised learning:** It is a machine learning method that uses both labeled and unlabeled data to train a model ibm +1. It sits between supervised learning and unsupervised learning, so it needs only a small amount of labeled data and a large amount of unlabeled data.

**Easy explanation**

**Batch ML**
Batch ML, or batch learning, means training a machine learning model on the entire dataset at once instead of updating it continuously with new data. It is also called offline learning because the model is trained first and then deployed after training.

***Easy meaning***

In batch ML, you collect the data, train the model on all of it together, and then use the trained model later for predictions. If new data comes in, the model usually has to be retrained again.

**How it works**

- First, all available training data is prepared.
- Then the model learns from that full dataset in one training phase.
- After training, the model is deployed and used for prediction.
- When the dataset changes a lot, retraining may be needed.

**Why it is used**
- Batch ML is useful when the data is mostly fixed and does not change too often.

- It is also helpful when you want stable training and can afford to process a large dataset before deployment.

**Advantages**

- Simple training process.
- Stable model updates.
- Good for static datasets.

**Limitations?**

- Not ideal for real-time data.
- Needs retraining when new data arrives.
- Can be slower for very large datasets.

Batch ML is a machine learning approach in which the model is trained on the full dataset at once, usually offline, and retrained later if new data is added

**Online machine learning** It is a method where the model learns continuously from new data instead of waiting for the whole dataset to be available. It is also called incremental learning or out-of-core learning.

**Easy meaning**
Think of it like studying one page at a time and updating your understanding after each page. The model keeps improving as new data arrives, so it is useful for fast-changing data such as customer behavior, stock prices, or streaming recommendations.

**How it works**

- The model gets data one example or a small chunk at a time.
- It updates its parameters immediately or frequently.
- It does not need to retrain from scratch every time new data comes in.

**Where it is used**
- Recommendation systems.
- Financial market analysis.
- Autonomous systems.
- Streaming analytics.

**Batch ML vs Online ML**
- Batch ML learns from the full dataset at once.
- Online ML learns step by step as data arrives.

Online machine learning is a type of machine learning in which the model learns from data continuously, updating itself as new data arrives

**Instance Based Learning:**

Instance-Based Learning is a "lazy" approach that stores all training data and makes predictions by comparing new points to the most similar stored examples using distance metrics. The classic k-NN algorithm classifies or predicts based on neighboring votes or averages. While simple, intuitive, and adaptive, it suffers from slow predictions, heavy memory use, and poor performance in high-dimensional data requiring feature scaling to work effectively.

**Model-Based Learning:**

Model-Based Learning is an "eager" approach where the algorithm analyzes the training data during a dedicated training phase to build a generalized, mathematical internal model (like a line, tree, or neural network) that captures underlying patterns. Once trained, this model is used to make quick predictions on new data without needing to reference the original dataset. While this makes predictions fast and memory-efficient, training can be computationally expensive, and the model's performance depends heavily on how well its assumptions match the actual data distribution.

**DIFFERENCE BTW IBN AND MBN**


| Feature | Instance-Based Learning | Model-Based Learning |
|---------|--------------------------|----------------------|
| **Nickname** | Lazy Learner | Eager Learner |
| **Training Phase** | Just stores the data (no computation) | Builds a generalized model from data |
| **When does the work happen?** | During prediction (expensive) | During training (expensive) |
| **Prediction Speed** | Slow (scans all data) | Fast (simple function call) |
| **Memory Usage** | High (stores entire dataset) | Low (only stores model parameters) |
| **Handles new data** | Easy — just add it to the store | Must retrain from scratch |
| **Assumptions** | None — non-parametric | Strong assumptions (e.g., linearity) |
| **Complex boundaries** | Can learn any shape | Limited by model capacity |
| **Dimensionality** | Fails in high dimensions (curse) | Can handle high dimensions well |
| **Interpretability** | Easy (nearest neighbors are explainable) | Varies (linear = easy, neural nets = black box) |
| **Example Algorithms** | k-NN, LSH, Ball Trees | Linear Regression, Decision Trees, Neural Networks, SVM |
| **Feature Scaling** | **Mandatory** (distances need it) | Usually optional (weights adjust) |
| **Missing Data** | Struggles (cannot compute distance) | Can often handle with imputation |
